```python
import os
import time
from typing import Any, List, Dict
```

### This code defines a very simple Python class called Agent. An Agent has a name and a run() method that accepts some data, prints a message, and returns the data unchanged.
### This creates a class named Agent. A class is like a blueprint for creating objects.
### This is a docstring. It describes what the class is intended to represent.
### __init__ is the constructor. It runs automatically when you create an Agent.
### It accepts input_data, which can be any type of Python object. Any normally comes from:
### The first Any means the input can be anything. The second Any, after ->, means the method can return anything
### Here, "Summarizer" is passed into name, and:tores it inside the object. So effectively:agent.name will be Summarize

```python
class Agent:
    """Simple agent that can perform a specific task"""
    
    def __init__(self, name: str):
        self.name = name
    
    def run(self, input_data: Any) -> Any:
        """Execute the agent's task"""
        print(f"Agent '{self.name}' processing: {str(input_data)[:30]}...")
        # This would be implemented by specific agents
        return input_data
```

##### Explanation of the ResearchAgent Class

The ResearchAgent class is a specialized agent that inherits from the previously defined Agent class. Its purpose is to simulate researching a given topic.

The run() method takes a research query as a string and prints a message showing the agent's name and the topic being researched. The time.sleep(0.5) statement pauses the program for 0.5 seconds to simulate the delay that might occur when making an API call or retrieving information from an external source.

After the simulated research is complete, the method returns a string containing sample research results. In this example, it reports that three key points were found about the requested topic.

Because ResearchAgent inherits from Agent, it can also use attributes defined in the parent class, such as self.name.

```python
class ResearchAgent(Agent):
    """Agent that simulates researching a topic"""
    
    def run(self, query: str) -> str:
        print(f"🔍 {self.name} researching: '{query}'")
        time.sleep(0.5)  # Simulate API call
        return f"Research results for '{query}': Found 3 key points about this topic."
```

##### Explanation of the SummarizerAgent Class

The SummarizerAgent class is a specialized agent that inherits from the previously defined Agent class. Its purpose is to simulate summarizing a piece of text.

The run() method takes a text input as a string and prints a message showing that the agent is summarizing the text. The time.sleep(0.5) statement pauses the program for 0.5 seconds to simulate the time needed to process the information.

The return statement creates a short summary from the input text. The expression text.split(':', 1)[1] splits the text at the first colon (:) and selects the part that comes after it. Then, [:30] keeps only the first 30 characters of that section, and ... is added at the end to make it look like a shortened summary.

Because SummarizerAgent inherits from Agent, it can also use attributes from the parent class, such as self.name.

```python
class SummarizerAgent(Agent):
    """Agent that summarizes information"""
    
    def run(self, text: str) -> str:
        print(f"📝 {self.name} summarizing text...")
        time.sleep(0.5)  # Simulate processing
        return f"Summary: {text.split(':', 1)[1][:30]}..."
```

##### Explanation of the FactCheckerAgent Class

The FactCheckerAgent class is a specialized agent that inherits from the previously defined Agent class. Its purpose is to simulate verifying or fact-checking a piece of information.

The run() method takes a text input as a string and prints a message showing that the agent is performing fact checking. The time.sleep(0.5) statement pauses the program for 0.5 seconds to simulate the time that might be required to verify the information.

The method returns a dictionary containing three pieces of information: the original text, an "accuracy" value set to "high", and "verified_claims" set to 3. This represents a simple simulated fact-checking result.

The -> Dict type hint indicates that the method returns a dictionary. Since FactCheckerAgent inherits from Agent, it can also use attributes from the parent class, such as self.name.

The final line, print("=== AGENTIC WORKFLOW DEMO ==="), displays a heading in the output. It is used to clearly mark the beginning of the agentic workflow demonstration when the program is executed.

```python
class FactCheckerAgent(Agent):
    """Agent that verifies information"""
    
    def run(self, text: str) -> Dict:
        print(f"✓ {self.name} fact checking...")
        time.sleep(0.5)  # Simulate verification
        return {
            "text": text,
            "accuracy": "high",
            "verified_claims": 3
        }

print("=== AGENTIC WORKFLOW DEMO ===")
```

```python
# Create agents
researcher = ResearchAgent("Research Assistant")
fact_checker = FactCheckerAgent("Fact Checker")
summarizer = SummarizerAgent("Summarizer")

print("\n🚀 Starting 'Information Processing' workflow\n")

# Initial input
query = "Agentic workflows in AI systems"

# Step 1: Research
research_results = researcher.run(query)
print(f"  → Output: {str(research_results)[:50]}...\n")

# Step 2: Fact check
fact_check_results = fact_checker.run(research_results)
print(f"  → Output: {str(fact_check_results)[:50]}...\n")

# Step 3: Summarize
summary = summarizer.run(fact_check_results["text"])
print(f"  → Output: {str(summary)[:50]}...\n")

print("✅ Workflow 'Information Processing' completed\n")

print("Final result:")
print(summary)

print("\nKey concepts demonstrated:")
print("1. Agents as components that perform specific tasks")
print("2. Workflow connecting agents in sequence")
print("3. Information flowing through the system")
print("4. Each agent transforming the data")
```
